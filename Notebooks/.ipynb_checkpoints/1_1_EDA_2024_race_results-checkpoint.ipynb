```python


import numbers
from re import S


def is_numeric(value):
    return isinstance(value, numbers.Number)


class SeasonPrior():
    def __init__(self, alpha: int = 1, previous_season: pd.DataFrame = None):
        """
        Initializes the BasicRaceForecast with a smoothing parameter alpha.
        :param alpha: Smoothing parameter for prior probability calculation.
        """
        self.alpha = alpha
        self.prior = {}
        self.previous_season = previous_season
        self.drivers = list(self.previous_season[['Driver', 'Constructor']
                                                 ].drop_duplicates().itertuples(index=False, name=None))
        self.constructors = self.previous_season['Constructor'].drop_duplicates(
        ).tolist()
        self.positions = [
            int(i) for i in self.previous_season["Pos."].unique() if i.isnumeric()]
        self.data = self.previous_season

    def fit_driver_prior(self):
        """_summary_

        Returns:
            _type_: _description_
        """
        self.prior = {driver: self.compute_prior(
            self.driver_counts()[driver]) for driver in self.drivers}
        return self.prior

    def fit_constructor_prior(self):
        """_summary_

        Returns:
            _type_: _description_
        """
        self.prior = {constructor: self.compute_prior(
            self.constructor_counts()[constructor]) for constructor in self.constructors}
        return self.prior

    def compute_prior(self, data):
        """
        Computes the prior probabilities for each driver based on their finishes.
        :param data: A dictionary where keys are driver names and values are lists of their finishes.
        """
        counts = pd.Series(data).value_counts().reindex(
            self.positions, fill_value=0)
        smoothed = (counts + self.alpha) / \
            (len(data) + self.alpha * len(self.positions))
        return smoothed.to_dict()

    def fit_season_prior(self, ):
        """
        Computes the prior probabilities for each driver based on their finishes.
        :param data: A dictionary where keys are driver names and values are lists of their finishes.
        """

        return self.compute_prior(self.default_counts())

    def clean_dataframe(self):
        """_summary_

        Args:
            df (pd.DataFrame): _description_
        """

        positions = [
            int(i) for i in self.previous_season["Pos."].unique() if i.isnumeric()]

        DN_positions_map = {'DNF': max(
            positions)+1, 'DNS': max(positions)+1, 'DSQ': max(positions)+1, 'DNQ': max(positions)+1, 'EX': max(positions)+1}
        self.previous_season['Pos_num'] = self.previous_season['Pos.'].replace(
            DN_positions_map).astype(int)

        return self.previous_season

    def driver_counts(self):
        """_summary_

        Returns:
            _type_: _description_
        """
        driver_counts = self.clean_dataframe().groupby(['Driver', 'Constructor'])[
            'Pos_num'].apply(list).to_dict()
        return driver_counts

    def default_counts(self):
        """_summary_

        Returns:
            _type_: _description_
        """
        default_counts = self.clean_dataframe()['Pos_num'].value_counts()
        return default_counts

    def constructor_counts(self):
        """_summary_

        Returns:
            _type_: _description_
        """
        constructor_counts = self.clean_dataframe().groupby('Constructor')[
            'Pos_num'].apply(list).to_dict()
        return constructor_counts


prior_2000 = SeasonPrior(alpha=.6, previous_season=results_2000)

prior_2000.fit_season_prior()


# # with open("/Users/bradkittrell/Projects/F1_forecast/F1_forecast/Data/processed/season_races_hist.json", 'rb') as f:
# #     season = json.load(f)
# # # convert_raw_processed(2004)

# for year in range(2000, 2024):
#     prior_save_year = f"{PROCESSED_PATH}/{year}/"
#     results_year = pd.read_csv(f"{prior_save_year}finishes.csv")
#     prior_year = SeasonPrior(alpha=.6, previous_season=results_year)
#     with open(f"{prior_save_year}prior_driver_{year}.json", 'w') as f:
#         json.dump(prior_year.fit_driver_prior(), f)
```

```python

from typing import Dict, List, Tuple


class RaceDataPreprocessor:
    @staticmethod
    def clean_positions(df: pd.DataFrame) -> pd.DataFrame:
        """
        Clean the 'Pos.' column and convert to numeric, mapping DNF/DNS/DSQ to a fallback value.
        """
        df = df.copy()
        raw_positions = df["Pos."].unique()
        numeric_positions = [int(i)
                             for i in raw_positions if str(i).isnumeric()]
        fallback_pos = max(numeric_positions) + 1

        DN_positions_map = {'DNF': fallback_pos,
                            'DNS': fallback_pos, 'DSQ': fallback_pos}
        df['Pos_num'] = df['Pos.'].replace(DN_positions_map).astype(int)

        return df

    @staticmethod
    def get_grouped_finishes(df: pd.DataFrame, groupby_cols: List[str]) -> Dict:
        """
        Groups the cleaned dataframe by specified columns and aggregates positions into lists.
        """
        df_clean = RaceDataPreprocessor.clean_positions(df)
        return df_clean.groupby(groupby_cols)['Pos_num'].apply(list).to_dict()

    @staticmethod
    def get_unique_positions(df: pd.DataFrame) -> List[int]:
        """
        Returns sorted list of unique numeric positions in the data.
        """
        return sorted([int(i) for i in df["Pos."].unique() if str(i).isnumeric()])

    def get_positions_count(df: pd.DataFrame) -> int:
        """
        Returns the number of unique positions in the data.
        """
        return len(self.get_unique_positions(df))


# class RollingBayesianForecaster:
#     def __init__(
#         self,
#         current_season: pd.DataFrame,
#         initial_prior: Dict,
#         constructor_prior: Dict[str, Dict[int, float]],
#         global_mean_prior: Dict[int, float],
#         window_size: int = 1)

#         self.prior = initial_prior.copy()
#         self.constructor_prior = constructor_prior
#         self.global_mean_prior = global_mean_prior
#         self.window_size = window_size
#         self.groupby_cols = groupby_cols
#         self.predictions = []

    # def run(self):
    #     unique_races = self.current_season['Race'].unique()

    #     for i in range(self.window_size, len(unique_races) + 1):
    #         race_subset = unique_races[i - self.window_size:i]
    #         window_df = self.current_season[self.current_season['Race'].isin(
    #             race_subset)]
    #         finishes = window_df.groupby(self.groupby_cols)[
    #             'Pos_num'].apply(list).to_dict()

    #         for key, new_data in finishes.items():
    #             old_prior = self._get_prior_or_fallback(key)
    #             updated_data = self._merge_data(old_prior, new_data)
    #             self.prior[key] = self.prior_builder.compute_prior(
    #                 updated_data)

    #         self.predictions.append(self.prior.copy())

    #     return self.predictions

    # def _merge_data(self, prior_dist: Dict[int, float], new_data: List[int]) -> List[int]:
    #     pseudo_obs = []
    #     total_pseudo = sum(prior_dist.values())
    #     for pos, prob in prior_dist.items():
    #         pseudo_obs.extend([pos] * int(prob * total_pseudo))
    #     return pseudo_obs + new_data

    # def _get_prior_or_fallback(self, key: Tuple[str, str]) -> Dict[int, float]:
    #     driver, constructor = key
    #     if key in self.prior:
    #         return self.prior[key]
    #     elif constructor in self.constructor_prior:
    #         return self.constructor_prior[constructor]
    #     else:
    #         return self.global_mean_prior


results_2001[results_2001['race_index'] <= 2]
```

```python
prior_window = 5              # past races used to form rolling prior
likelihood_window = 1         # recent races used to form likelihood
burn_in_index = 4             # use previous season prior until this race_index
groupby_cols = ['Driver', 'Constructor']  # grouped model


class PriorBuilder:
    def __init__(self, alpha: float = 1.0, positions: List[int] = None):
        self.alpha = alpha
        self.positions = positions

    def compute_prior(self, finishes: List[int]) -> Dict[int, float]:
        """
        Compute smoothed prior distribution over finishing positions.
        """
        counts = pd.Series(finishes).value_counts().reindex(
            self.positions, fill_value=0)
        smoothed = (counts + self.alpha) / \
            (len(finishes) + self.alpha * len(self.positions))
        return smoothed.to_dict()


class BayesianRollingForecasterV2:
    def __init__(
        self,
        current_season: pd.DataFrame,
        prior_season: pd.DataFrame,
        alpha: float = 1.0,
        groupby_cols: List[str] = ['Driver', 'Constructor'],
        prior_window: int = 3,
        likelihood_window: int = 1,
        burn_in_index: int = 4
    ):
        self.current_season = RaceDataPreprocessor.clean_positions(
            current_season)
        self.prior_season = RaceDataPreprocessor.clean_positions(prior_season)
        self.groupby_cols = groupby_cols
        self.prior_window = prior_window
        self.likelihood_window = likelihood_window
        self.burn_in_index = burn_in_index

        # Infer all unique positions across both datasets
        all_positions = pd.concat([
            self.current_season['Pos_num'],
            self.prior_season['Pos_num']
        ])
        self.positions = sorted(all_positions.unique())

        # Now we can create a PriorBuilder instance safely
        self.prior_builder = PriorBuilder(
            alpha=alpha, positions=self.positions)

        # Store posterior updates over the season
        self.posterior_history = []

        # Compute initial prior from the previous season
        self.initial_prior = RaceDataPreprocessor.get_grouped_finishes(
            self.prior_season, self.groupby_cols)
        self.initial_prior = {
            k: self.prior_builder.compute_prior(v)
            for k, v in self.initial_prior.items()
        }

    def run(self) -> List[Dict]:
        race_indices = sorted(self.current_season['race_index'].unique())

        for r in race_indices:
            if r <= self.burn_in_index:
                prior = self.initial_prior
                likelihood_df = self._get_window_df(
                    r - self.likelihood_window, r)
            else:
                prior_df = self._get_window_df(
                    r - self.likelihood_window - self.prior_window, r - self.likelihood_window)
                likelihood_df = self._get_window_df(
                    r - self.likelihood_window, r)
                prior = self._compute_prior_from_df(prior_df)

            likelihood = self._compute_likelihood_from_df(likelihood_df)
            posterior = self._update_posterior(prior, likelihood)
            self.posterior_history.append(posterior)

        return self.posterior_history

    def _get_window_df(self, start_index: int, end_index: int) -> pd.DataFrame:
        return self.current_season[
            self.current_season['race_index'].between(
                start_index + 1, end_index)
        ]

    def _compute_prior_from_df(self, df: pd.DataFrame) -> Dict[Tuple, Dict[int, float]]:
        grouped = df.groupby(self.groupby_cols)[
            'Pos_num'].apply(list).to_dict()
        return {
            k: self.prior_builder.compute_prior(v)
            for k, v in grouped.items()
        }

    def _compute_likelihood_from_df(self, df: pd.DataFrame) -> Dict[Tuple, List[int]]:
        return df.groupby(self.groupby_cols)['Pos_num'].apply(list).to_dict()

    def _update_posterior(self, prior_dist: Dict, likelihood_data: Dict) -> Dict:
        updated = {}

        for key in set(prior_dist.keys()).union(likelihood_data.keys()):
            prior = prior_dist.get(
                key, {pos: self.prior_builder.alpha for pos in self.positions})
            likelihood = likelihood_data.get(key, [])

            # Convert prior dist to pseudo-counts
            pseudo_counts = {
                pos: prior.get(pos, 0) *
                self.prior_builder.alpha * len(self.positions)
                for pos in self.positions
            }

            for pos in likelihood:
                pseudo_counts[pos] += 1

            # Normalize
            total = sum(pseudo_counts.values())
            updated[key] = {pos: count / total for pos,
                            count in pseudo_counts.items()}

        return updated


def score_posterior_against_actuals(
    posterior: Dict[Tuple[str, str], Dict[int, float]],
    actual_race_df: pd.DataFrame,
    groupby_cols: List[str] = ['Driver', 'Constructor'],
    pos_col: str = 'Pos_num'
) -> Dict[str, float]:
    """
    Scores the posterior predictions against actual outcomes using log-likelihood.

    Args:
        posterior: Dict mapping (Driver, Constructor) to position distributions.
        actual_race_df: DataFrame with true results for a single race.
        groupby_cols: Columns used to group entities (usually Driver/Constructor).
        pos_col: Column in actual_race_df that contains true positions.

    Returns:
        Dict with:
            - 'mean_log_score': average log-likelihood across drivers
            - 'per_driver_log_scores': individual scores
    """
    log_scores = []
    per_driver_scores = {}

    for _, row in actual_race_df.iterrows():
        key = tuple(row[col] for col in groupby_cols)
        actual_pos = int(row[pos_col])

        if key not in posterior:
            # No prediction — treat as uniform or skip
            continue

        predicted_dist = posterior[key]
        # small epsilon to avoid log(0)
        prob = predicted_dist.get(actual_pos, 1e-8)

        log_prob = np.log(prob)
        log_scores.append(log_prob)
        per_driver_scores[str(key)] = log_prob

    if not log_scores:
        return {
            'mean_log_score': float('-inf'),
            'per_driver_log_scores': per_driver_scores
        }

    return {
        'mean_log_score': np.mean(log_scores),
        'per_driver_log_scores': per_driver_scores
    }


class RaceDataPreprocessor:
    @staticmethod
    def clean_positions(df: pd.DataFrame) -> pd.DataFrame:
        """
        Cleans the 'Pos.' column, converting it into a numeric 'Pos_num' column.
        Known non-finish statuses are mapped to max finish position + 1.
        """
        df = df.copy()

        # Get numeric positions
        raw_positions = df["Pos."].astype(str).unique()
        numeric_positions = [int(p) for p in raw_positions if p.isnumeric()]
        if not numeric_positions:
            raise ValueError(
                "No numeric race positions found in 'Pos.' column.")

        max_position = max(numeric_positions)
        fallback_position = max_position + 1

        # Define known non-numeric status codes (feel free to extend this list)
        non_finish_map = {
            'DNF': fallback_position,
            'DNS': fallback_position,
            'DSQ': fallback_position,
            'DNQ': fallback_position,
            'NC': fallback_position,    # Not Classified (optional)
            'Ret': fallback_position,   # Retired (optional)
        }

        # Convert all to string for mapping
        df["Pos."] = df["Pos."].astype(str)

        # Apply mapping
        df['Pos_num'] = df["Pos."].replace(non_finish_map)

        # Convert remaining numerics
        df['Pos_num'] = df['Pos_num'].apply(
            lambda x: int(x) if str(x).isnumeric() else fallback_position
        )

        return df

    def score_against_actual(self, race_index: int) -> Dict:
        posterior = self.posterior_history[race_index - 1]
        actual_df = self.current_season[self.current_season['race_index'] == race_index]
        return score_posterior_against_actuals(posterior, actual_df, self.groupby_cols)

    @staticmethod
    def get_grouped_finishes(df: pd.DataFrame, groupby_cols: List[str]) -> Dict:
        """
        Groups the cleaned dataframe by specified columns and aggregates positions into lists.
        """
        df_clean = RaceDataPreprocessor.clean_positions(df)
        return df_clean.groupby(groupby_cols)['Pos_num'].apply(list).to_dict()

    @staticmethod
    def get_unique_positions(df: pd.DataFrame) -> List[int]:
        """
        Returns sorted list of unique numeric positions in the data.
        """
        return sorted([int(i) for i in df["Pos."].unique() if str(i).isnumeric()])


prior_2000 = PriorBuilder(
    alpha=1.0, positions=RaceDataPreprocessor.get_unique_positions(results_2000))
compute_prior(RaceDataPreprocessor.get_grouped_finishes(df=results_2000,
                                                        groupby_cols=['Driver', 'Constructor']))
```
